```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/Srujanmp1366/flyrank-internship"
REPO_DIR = "flyrank-internship"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)

print("Current directory:", os.getcwd())
print(os.listdir("data/raw"))
```

```python
import pandas as pd

# Load the dataset
url = 'https://raw.githubusercontent.com/Srujanmp1366/flyrank-internship/main/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(url)

# 1. Print the exact column names so you know what you are working with
print("Available columns:", df.columns.tolist())

# 2. Dynamically grab the right column names (whether they have a suffix or not)
imp_col = [c for c in df.columns if 'impression' in c.lower()][0]
click_col = [c for c in df.columns if 'click' in c.lower()][0]
pos_col = [c for c in df.columns if 'position' in c.lower()][0]

# Calculate CTR dynamically just in case it's missing entirely
ctr_col = 'ctr_calculated'
df[ctr_col] = (df[click_col] / df[imp_col]).fillna(0)

# 3. Find the "Wasted Visibility" Archetype
imp_90 = df[imp_col].quantile(0.90)
ctr_10 = df[ctr_col].quantile(0.10)
high_imp_low_ctr = df[(df[imp_col] > imp_90) & (df[ctr_col] <= ctr_10)]

# 4. Find the "Hidden Gems" Archetype
imp_50 = df[imp_col].quantile(0.50)
ctr_90 = df[ctr_col].quantile(0.90)
hidden_gems = df[(df[imp_col] < imp_50) & (df[ctr_col] > ctr_90)]

# 5. Find the "High Rank, Zero Value" Archetype
stale_front_page = df[(df[pos_col] < 10) & (df[click_col] == 0)]

print(f"\n--- Why We Need Lane 3 Clustering ---")
print(f"Wasted Visibility (Top 10% Impressions, Bottom 10% CTR): {len(high_imp_low_ctr)} pages")
print(f"Hidden Gems (Bottom 50% Impressions, Top 10% CTR): {len(hidden_gems)} pages")
print(f"Front Page, Zero Clicks (Ranked < 10, but 0 Clicks): {len(stale_front_page)} pages")
```
